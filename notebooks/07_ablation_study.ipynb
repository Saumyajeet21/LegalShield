Inspect the estimate, then execute the opted-in pilot locally after ensuring reviewed inputs, a checkpoint and an approved budget. This notebook consumes actual predictions and plots paired and aggregate outcomes.

Sources: [ContractNLI](https://aclanthology.org/2021.findings-emnlp.164/), [DeBERTa-v3](https://arxiv.org/abs/2111.09543) and [dataset preparation](https://stanfordnlp.github.io/contract-nli/). Persist outputs from the Colab VM before ending the session.

```python
import os, subprocess, sys, json
from pathlib import Path
# Upload or mount the upgrade branch, then point this at its path in Colab.
PROJECT = Path('/content/LegalShield')
assert (PROJECT / 'research' / 'requirements.txt').exists(), 'Set PROJECT to the checked-out upgrade branch'
os.chdir(PROJECT)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'research/requirements.txt'], check=True)
subprocess.run([sys.executable, '-m', 'pip', 'install', '-r', 'backend/requirements-verification.txt'], check=True)
import torch
print(torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU runtime')
def run(module, *args):
    subprocess.run([sys.executable, '-m', module, *map(str, args)], check=True)
```

```python
data = Path('artifacts/data/laven-pilot.jsonl')
run('research.run_pilot', data, 'artifacts/ablation-predictions.jsonl', '--limit', '8', '--systems', 'M3', 'M7', 'without_authority', 'without_neural', 'without_symbolic', 'without_uncertainty', 'without_regeneration')
predictions = Path('artifacts/ablation-predictions.jsonl')
if predictions.exists():
    run('research.experiments', data, predictions, 'artifacts/ablation-metrics.json')
    run('research.plots', 'comparisons', 'artifacts/ablation-metrics.json', 'artifacts/ablation-figures')
```
